```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

def main():
    model.train(
        data="dataset.yaml",
        epochs=80,
        imgsz=640,
        batch=16,
        device=0,   # GPU
        workers=8,
        lr0=0.001,
        patience=20,
        project="runs",
        name="drone_detector"
    )

if __name__ == "__main__":
    main()
```
